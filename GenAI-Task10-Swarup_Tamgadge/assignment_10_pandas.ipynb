# Task 1 — Pandas Series Basics

```python

import pandas as pd

marks = [78, 85, 90, 66, 72]

series = pd.Series(marks)

print("Series Values:")
print(series.values)

print("\nIndex:")
print(series.index)

print("\nData Type:")
print(series.dtype)

print("\nFirst Element:")
print(series.iloc[0])

print("\nLast Two Elements:")
print(series.iloc[-2:])
```

# Task 2 — Mathematical Operations on Series

```python
import pandas as pd

marks = pd.Series([78, 85, 90, 66, 72])

print("Original Marks:")
print(marks)

print("\nAfter Adding 5 Grace Marks:")
print(marks + 5)

print("\nAfter Subtracting 2 Marks:")
print(marks - 2)

print("\nAfter Multiplying by 1.05:")
print(marks * 1.05)

print("\nAfter Dividing by 2:")
print(marks / 2)
```

# Task 3 — Python Functionalities on Series

```python
import pandas as pd

marks = pd.Series([78, 85, 90, 66, 72])

print("Maximum Marks:", marks.max())
print("Minimum Marks:", marks.min())
print("Sum of Marks:", marks.sum())
print("Mean Marks:", marks.mean())

passed = marks.apply(lambda x: x >= 70)

print("\nPass Status:")
print(passed)

print("\nNumber of Students Passed:", passed.sum())
```

# Task 4 — Create a DataFrame

# Task 5 — Important DataFrame Functions

```python

import pandas as pd

students = {
    'Name': ['Amit', 'Neha', 'Rahul', 'Sneha', 'Pooja'],
    'Marks': [78, 85, 90, 66, 72],
    'Subject': ['Math', 'Math', 'Science', 'Science', 'Math']
}

df = pd.DataFrame(students)

print("DataFrame Information:")
df.info()

print("\nStatistical Description:")
print(df.describe())

print("\nFirst 5 Rows:")
print(df.head())

print("\nLast 5 Rows:")
print(df.tail())

print("\nSorted by Marks in Descending Order:")
sorted_df = df.sort_values(by='Marks', ascending=False)
print(sorted_df)

print("\nSorted DataFrame with Reset Index:")
sorted_df = sorted_df.reset_index(drop=True)
print(sorted_df)
```

# Task 6 — Filtering & Conditional Selection

```python
import pandas as pd

students = {
    'Name': ['Amit', 'Neha', 'Rahul', 'Sneha', 'Pooja'],
    'Marks': [78, 85, 90, 66, 72],
    'Subject': ['Math', 'Math', 'Science', 'Science', 'Math']
}

df = pd.DataFrame(students)

print("Students Scoring More Than 75:")
print(df[df['Marks'] > 75])

print("\nStudents Belonging to Math:")
print(df[df['Subject'] == 'Math'])

average = df['Marks'].mean()

print("\nAverage Marks:", average)
print("\nStudents Scoring Above Average:")
print(df[df['Marks'] > average])

print("\nStudents Who Failed:")
print(df[df['Marks'] < 70])
```

# Task 7 — Grouping & Basic Analysis

```python
import pandas as pd

students = {
    'Name': ['Amit', 'Neha', 'Rahul', 'Sneha', 'Pooja'],
    'Marks': [78, 85, 90, 66, 72],
    'Subject': ['Math', 'Math', 'Science', 'Science', 'Math']
}

df = pd.DataFrame(students)

print("Average Marks per Subject:")
print(df.groupby('Subject')['Marks'].mean())

print("\nNumber of Students per Subject:")
print(df.groupby('Subject')['Name'].count())

print("\nMaximum Marks per Subject:")
print(df.groupby('Subject')['Marks'].max())
```

# Task 8 — Pandas Plotting (Simple Graphs)

```python
import pandas as pd

students = {
    'Name': ['Amit', 'Neha', 'Rahul', 'Sneha', 'Pooja'],
    'Marks': [78, 85, 90, 66, 72],
    'Subject': ['Math', 'Math', 'Science', 'Science', 'Math']
}

df = pd.DataFrame(students)

# 1. Bar graph of student names vs marks
df.plot(x='Name', y='Marks', kind='bar', title='Student Marks')

# 2. Line graph of marks
df['Marks'].plot(kind='line', title='Marks Line Graph', marker='o')

# 3. Histogram of marks
df['Marks'].plot(kind='hist', title='Marks Histogram', bins=5)

import matplotlib.pyplot as plt
plt.show()
```
