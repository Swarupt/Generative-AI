# Task 1: Line Plot (Sales Trend)

```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv('sales_data (1).csv')
df['Sale_Date'] = pd.to_datetime(df['Sale_Date'])
monthly_sales = df.groupby(df['Sale_Date'].dt.to_period('M'))[
    'Sales_Amount'
].sum()

plt.plot(monthly_sales.index.astype(str), monthly_sales.values, marker='o')
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.show()
```

# Task 2: Scatter Plot

```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv('sales_data (1).csv')

plt.scatter(df['Quantity_Sold'], df['Sales_Amount'], color='blue', alpha=0.5)
plt.title('Quantity Sold vs Sales Amount')
plt.xlabel('Quantity Sold')
plt.ylabel('Sales Amount')
plt.show()
```

# Task 3: Vertical & Horizontal Bar Plot

```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv('sales_data (1).csv')
category_sales = df.groupby('Product_Category')['Sales_Amount'].sum()

# Vertical Bar Chart
plt.bar(category_sales.index, category_sales.values)
plt.title('Sales by Product Category (Vertical)')
plt.xlabel('Product Category')
plt.ylabel('Total Sales')
plt.show()

# Horizontal Bar Chart
plt.barh(category_sales.index, category_sales.values)
plt.title('Sales by Product Category (Horizontal)')
plt.xlabel('Total Sales')
plt.ylabel('Product Category')
plt.show()
```

# Task 4: Multiple Bar Plot

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

df = pd.read_csv('sales_data (1).csv')
df['Year'] = pd.to_datetime(df['Sale_Date']).dt.year
pivot_sales = df.pivot_table(
    index='Product_Category',
    columns='Year',
    values='Sales_Amount',
    aggfunc='sum',
).fillna(0)

x = np.arange(len(pivot_sales.index))
width = 0.35

plt.bar(x - width / 2, pivot_sales[2023], width, label='2023')
plt.bar(x + width / 2, pivot_sales[2024], width, label='2024')
plt.title('Sales Comparison by Category across Years')
plt.xlabel('Product Category')
plt.ylabel('Total Sales')
plt.xticks(x, pivot_sales.index)
plt.legend()
plt.show()
```

# Task 5: Stacked Bar Chart

```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv('sales_data (1).csv')
pivot_region = df.pivot_table(
    index='Product_Category', columns='Region', values='Sales_Amount', aggfunc='sum'
).fillna(0)

bottom = None
for region in pivot_region.columns:
  plt.bar(
      pivot_region.index, pivot_region[region], bottom=bottom, label=region
  )
  bottom = (
      pivot_region[region]
      if bottom is None
      else bottom + pivot_region[region]
  )

plt.title('Stacked Bar Chart - Sales by Category and Region')
plt.xlabel('Product Category')
plt.ylabel('Total Sales')
plt.legend()
plt.show()
```

# Task 6: Histogram (Numeric Distribution)

```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv('sales_data (1).csv')

plt.hist(df['Sales_Amount'], bins=10, edgecolor='black')
plt.title('Sales Amount Distribution')
plt.xlabel('Sales Amount')
plt.ylabel('Frequency')
plt.show()
```

# Task 7: Pie Chart (Market Share)

```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv('sales_data (1).csv')
category_sales = df.groupby('Product_Category')['Sales_Amount'].sum()

plt.pie(
    category_sales.values,
    labels=category_sales.index,
    autopct='%1.1f%%',
    startangle=90,
)
plt.title('Market Share by Product Category')
plt.show()
```
