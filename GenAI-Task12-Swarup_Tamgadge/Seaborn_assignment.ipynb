# Task 1: Relational Plot (sns.relplot)

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

df = pd.read_csv('sales_data (1).csv')

# 1 & 2. Relational plot with hue using a categorical column
sns.relplot(
    data=df,
    x='Quantity_Sold',
    y='Sales_Amount',
    hue='Product_Category',
    kind='scatter',
)
plt.title('Relational Plot with Hue')
plt.show()

# 3. Same plot using scatter style (markers/style)
sns.relplot(
    data=df,
    x='Quantity_Sold',
    y='Sales_Amount',
    hue='Product_Category',
    style='Product_Category',
    kind='scatter',
)
plt.title('Relational Plot with Scatter Style')
plt.show()
```

# Task 2: Line Plot as Scatter & Facet (sns.lineplot & FacetGrid)

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

df = pd.read_csv('sales_data (1).csv')

# 1. Line plot using sns.lineplot()
sns.lineplot(data=df, x='Quantity_Sold', y='Sales_Amount')
plt.title('Standard Line Plot')
plt.show()

# 2. Scatter-style line plot (markers on line)
sns.lineplot(
    data=df, x='Quantity_Sold', y='Sales_Amount', marker='o', linestyle='none'
)
plt.title('Scatter-Style Line Plot')
plt.show()

# 3. Faceting using col based on a categorical column
g = sns.FacetGrid(df, col='Region')
g.map(sns.lineplot, 'Quantity_Sold', 'Sales_Amount', marker='o')
plt.show()
```

# Task 3: Distribution Plots

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

df = pd.read_csv('sales_data (1).csv')

# 1. Histogram
sns.histplot(df['Sales_Amount'], kde=False)
plt.title('Histogram of Sales Amount')
plt.show()

# 2. KDE plot
sns.kdeplot(df['Sales_Amount'], fill=True)
plt.title('KDE Plot of Sales Amount')
plt.show()

# 3. Rug plot
sns.rugplot(df['Sales_Amount'])
plt.title('Rug Plot of Sales Amount')
plt.show()

# 4. Combine Histogram + KDE in a single plot
sns.histplot(df['Sales_Amount'], kde=True, stat='density')
plt.title('Histogram + KDE Plot')
plt.show()
```

# Task 4: Bivariate Distribution Plots

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

df = pd.read_csv('sales_data (1).csv')

# 1. Bivariate histogram
sns.histplot(
    data=df, x='Quantity_Sold', y='Sales_Amount', cbar=True, bins=10
)
plt.title('Bivariate Histogram')
plt.show()

# 2. Bivariate KDE plot
sns.kdeplot(data=df, x='Quantity_Sold', y='Sales_Amount', fill=True)
plt.title('Bivariate KDE Plot')
plt.show()
```

# Task 5: Matrix Plots

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

df = pd.read_csv('sales_data (1).csv')

# 1. Pair plot using sns.pairplot()
sns.pairplot(df[['Sales_Amount', 'Quantity_Sold', 'Unit_Price', 'Unit_Cost']])
plt.show()

# 2. Heatmap of correlation matrix
corr = df[
    ['Sales_Amount', 'Quantity_Sold', 'Unit_Price', 'Unit_Cost', 'Discount']
].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Correlation Matrix Heatmap')
plt.show()
```

# Task 6: Categorical Plots

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

df = pd.read_csv('sales_data (1).csv')

# 1. Bar plot
sns.barplot(data=df, x='Product_Category', y='Sales_Amount')
plt.title('Categorical Bar Plot')
plt.show()

# 2. Box plot
sns.boxplot(data=df, x='Product_Category', y='Sales_Amount')
plt.title('Categorical Box Plot')
plt.show()

# 3. Violin plot
sns.violinplot(data=df, x='Product_Category', y='Sales_Amount')
plt.title('Categorical Violin Plot')
plt.show()

# 4. Count plot
sns.countplot(data=df, x='Product_Category')
plt.title('Categorical Count Plot')
plt.show()
```

# Task 7: Regression Plots

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

df = pd.read_csv('sales_data (1).csv')

# 1. Regression plot (regplot)
sns.regplot(data=df, x='Quantity_Sold', y='Sales_Amount')
plt.title('Regression Plot')
plt.show()

# 2. Implot with hue
sns.lmplot(
    data=df,
    x='Quantity_Sold',
    y='Sales_Amount',
    hue='Product_Category',
    aspect=1.5,
)
plt.title('LMPlot with Hue')
plt.show()
```

# Task 8: Multi-Plots & Figure-Level Plots

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

df = pd.read_csv('sales_data (1).csv')

# 1. FacetGrid with numerical x, numerical y, and categorical col
g = sns.FacetGrid(df, col='Product_Category', height=4, aspect=0.8)
g.map(sns.scatterplot, 'Quantity_Sold', 'Sales_Amount')
plt.show()

# 2. Multi-plot dashboards using relplot, catplot, displot
# Relational Dashboard
g_rel = sns.relplot(
    data=df,
    x='Quantity_Sold',
    y='Sales_Amount',
    hue='Region',
    col='Product_Category',
    kind='scatter',
)
plt.show()

# Categorical Dashboard
g_cat = sns.catplot(
    data=df, x='Product_Category', y='Sales_Amount', hue='Region', kind='bar'
)
plt.show()

# Distribution Dashboard
g_dis = sns.displot(
    data=df, x='Sales_Amount', hue='Product_Category', kind='kde', fill=True
)
plt.show()
```
